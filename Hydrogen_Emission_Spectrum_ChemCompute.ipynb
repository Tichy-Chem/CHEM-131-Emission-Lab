```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider

def bohr_atom(max_n=6):
    # Bohr energy levels in eV
    n = np.arange(1, max_n + 1)
    E = -13.6 / n**2

    fig, ax = plt.subplots(figsize=(8, 6))

    # Draw energy levels
    for ni, Ei in zip(n, E):
        ax.hlines(Ei, 0.2, 0.8, linewidth=3)
        ax.text(0.83, Ei, f"n = {ni}", va="center", fontsize=12)

    # Draw the nucleus symbolically
    ax.scatter(0.5, -14.5, s=300)
    ax.text(0.5, -14.5, "+", ha="center", va="center",
            fontsize=18)

    ax.set_xlim(0, 1)
    ax.set_ylim(-15, 1)

    ax.set_ylabel("Energy (eV)", fontsize=12)
    ax.set_xticks([])

    ax.set_title("Bohr Model Energy Levels")

    # Show energy gaps
    if max_n > 1:
        for i in range(len(E) - 1):
            gap = E[i + 1] - E[i]
            ax.text(
                0.05,
                (E[i] + E[i + 1]) / 2,
                f"ΔE = {gap:.2f} eV",
                fontsize=9,
                va="center"
            )

    plt.show()


interact(
    bohr_atom,
    max_n=IntSlider(
        value=4,
        min=1,
        max=10,
        step=1,
        description="Highest n:"
    )
);
```

```python
import numpy as np
import matplotlib.pyplot as plt
import math

# Constants
R_H = 2.18e-18  # Rydberg energy in Joules (approximate)

# Quantum numbers n = 1 to 10
n = np.arange(1, 11)

# Calculate energies
E = -R_H / n**2

# Plot E vs n
plt.figure(figsize=(10,5))

plt.subplot(1, 2, 1)
plt.plot(n, E, 'o-', color='blue')
plt.title('Energy E vs. n')
plt.xlabel('n')
plt.ylabel('Energy (Joules)')
plt.grid(True)

# Plot E vs n^2
plt.subplot(1, 2, 2)
plt.plot(n**2, E, 'o-', color='red')
plt.title('Energy E vs. n²')
plt.xlabel('n²')
plt.ylabel('Energy (Joules)')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Rydberg constant (J)
R_H = 2.18e-18

# Quantum numbers n = 1 through 6
n = np.arange(1, 7)

# Define x = 1/n^2  and y = E_n
x = 1 / (n**2)
E = -R_H * x

plt.figure()
plt.plot(x, E, marker='o')
plt.xlabel('1 / n²')
plt.ylabel('Energy (J)')
plt.title('Energy vs 1/n² for Hydrogen Atom')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# --------------------------------------------------
# USER INPUT: Enter any hydrogen transitions here.
# Each pair is written as (n_initial, n_final).
# --------------------------------------------------

n_trans = [(__, __), (__, __)]


# Physical constants

R_H = 2.179872e-18       # Rydberg energy, J
h = 6.62607015e-34       # Planck's constant, J·s
c = 299792458            # Speed of light, m/s


# Calculate the energy of an electron at energy level n

def hydrogen_energy(n):
    return -R_H / n**2


# Calculate ΔE, frequency, and wavelength for one transition

def calculate_transition(n_initial, n_final):

    # Check that both n values are positive whole numbers
    if not isinstance(n_initial, int) or not isinstance(n_final, int):
        raise ValueError("Each n value must be a whole number.")

    if n_initial < 1 or n_final < 1:
        raise ValueError("Each n value must be 1 or greater.")

    if n_initial == n_final:
        raise ValueError(
            "The initial and final energy levels must be different."
        )

    E_initial = hydrogen_energy(n_initial)
    E_final = hydrogen_energy(n_final)

    # Energy change of the atom
    delta_E = E_final - E_initial

    # Photon frequency and wavelength use the magnitude of ΔE
    frequency = abs(delta_E) / h
    wavelength_m = c / frequency
    wavelength_nm = wavelength_m * 10**9

    # Identify the type of transition
    if delta_E < 0:
        transition_type = "Emission"
    else:
        transition_type = "Absorption"

    return {
        "n initial": n_initial,
        "n final": n_final,
        "Transition": transition_type,
        "ΔE (J)": f"{delta_E:.3E}",
        "Frequency (s⁻¹)": f"{frequency:.3E}",
        "Wavelength (nm)": f"{wavelength_nm:.1f}"
    }


# Calculate every transition entered above

results = []

for n_initial, n_final in n_trans:
    result = calculate_transition(n_initial, n_final)
    results.append(result)


# Create and display the results table

df = pd.DataFrame(results)

fig_height = max(2.2, 0.55 * (len(df) + 2))
fig, ax = plt.subplots(figsize=(12, fig_height))

ax.axis("off")

table = ax.table(
    cellText=df.values,
    colLabels=df.columns,
    cellLoc="center",
    colLoc="center",
    loc="center"
)

table.auto_set_font_size(False)
table.set_fontsize(11)
table.scale(1, 1.6)

# Format the header row
for column_number in range(len(df.columns)):
    header_cell = table[(0, column_number)]
    header_cell.set_facecolor("#D7EAFC")
    header_cell.set_text_props(weight="bold")

plt.title(
    "Hydrogen Electronic Transitions",
    fontsize=14,
    fontweight="bold",
    pad=16
)

plt.tight_layout()
plt.show()
```

```python
import ipywidgets as widgets
import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display, clear_output # Import clear_output

# Create a table input widget for color and wavelength
input_table = widgets.GridspecLayout(5, 2)
input_table[0, 0] = widgets.Label("Color of line")
input_table[0, 1] = widgets.Label("Wavelength (nm)")

for i in range(1, 5):
    input_table[i, 0] = widgets.Text()
    input_table[i, 1] = widgets.FloatText()

submit_btn = widgets.Button(description='Submit Wavelengths', button_style='primary')
output = widgets.Output()

display(widgets.VBox([input_table, submit_btn, output]))

# Make table with values for Color, Wavelength (nm)
def make_table(b):
    global wl_nm # Make wl_nm globally accessible
    with output: # Use the output widget to display results
        clear_output() # Clear previous output
        data = []
        # Create a list of input wavelengths
        wl_nm = [input_table[i, 1].value for i in range(1, 5)] # Populate wl_nm list

        #print("Input wavelengths (wl_nm):", wl_nm) # Print wl_nm for debugging

        for i in range(1, 5):
            color = input_table[i, 0].value
            wavelength = input_table[i, 1].value
            if color and wavelength: # Only add if both are provided
                data.append([color, int(round(wavelength))]) # Round wavelength to integer

        if data: # Only create table if there is data
            fig, ax = plt.subplots(1, 1)
            df = pd.DataFrame(data, columns=["Color of line", "Wavelength (nm)"])
            ax.axis('off')
            ax.set_title("Figure 3: Wavelengths of colored light (nm)", fontsize=36, y=1.2, loc="center")
            table = ax.table(cellText=df.values,
                             colLabels=df.columns,
                             loc="center")
            table.set_fontsize(30)
            table.scale(5, 5)
            plt.show()
        else:
            print("Please enter both color and wavelength for at least one row.")


submit_btn.on_click(make_table)
```
